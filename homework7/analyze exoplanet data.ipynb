```python
# Planet Name - pl_name: Planet name most commonly used in the literature. This means what the planet is called commonly.
# Host Name - hostname: Stellar name most commonly used in the literature. This means what the solar system is called commonly.
# Discovery Method - discoverymethod: Method by which the planet was first identified. This means the method in which they found the planet for the first time.
# Spectral Type - st_spectype: Classification of the star based on their spectral characteristics following the Morgan-Keenan system. This means how stars are classified based on their spectral characteristics.
# Equilibrium Temperature [K] - pl_eqt: The equilibrium temperature of the planet as modeled by a black body heated only by its host star, or for directly imaged planets, the effective temperature of the planet required to match the measured luminosity if the planet were a black body. This means the estimted temperature if a celestial body absorbed sunlight and re-radiated heat perfectly without any atmosphere or internal heating.
# Planet Mass or Mass*sin(i) [Jupiter Mass] - pl_bmassjBest planet mass estimate available, in order of preference: Mass, M*sin(i)/sin(i), or M*sin(i), depending on availability, and measured in Jupiter masses. This means the mass of a celestial body measured in units of Jupiter's mass.
```

```python
# Transit - Star dims slightly when planet passes in front, helping astronomers detect it.
# Radial Velocity - The small movements of a planet's star is caused by the planet's gravity, helping astronomers detect it.
# Microlensing - Gravitational bending of light from a distant background star helps detect a planet.
# Imaging - Directly capturing light from the planet separately from its star.
# Transit Timing Variations - Planet does not travel in front of its star at perfectly timed intervals.
# Eclipse Timing Variations - Changes in the timing of eclipses in a star system helps detect a planet.
# Disk Kinematics - Planets forming inside of protoplanetary disks create localized disturbances that can be detected.
```

```python
plt.figure(figsize=(8, 6))

for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]
    plt.scatter(subset["pl_bmassj"], subset["pl_eqt"], label=method)

plt.title("Planet Mass vs. Equilibrium Temperature")
plt.xlabel("Planet Mass (Jupiter Masses)")
plt.ylabel("Equilibrium Temperature (K)")
plt.legend(title="Discovery Method")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]
    plt.scatter(np.log(subset["pl_bmassj"]), np.log(subset["pl_eqt"]), label=method)

plt.title("Planet Mass vs. Equilibrium Temperature (Log Scale)")
plt.xlabel("Log of Planet Mass")
plt.ylabel("Log of Equilibrium Temperature")
plt.legend(title="Discovery Method")
plt.show()
```

```python
# They are more likely to be massive. This makes sense as spotting planets that are massive is easier than finding smaller planets.
```

```python
mass_filtered = df2[df2["pl_bmassj"] > 5]
coldest_planet = mass_filtered["pl_eqt"].idxmin()
print(df2.loc[coldest_planet])
```

```python
print("Planet Name:", df2.loc[coldest_planet, "pl_name"])
print("Index:", coldest_planet)
print("Equilibrium Temperature:", df2.loc[coldest_planet, "pl_eqt"])
print("Planet Mass:", df2.loc[coldest_planet, "pl_bmassj"])
print("Star Spectral Type:", df2.loc[coldest_planet, "st_spectype"])
```

```python
plt.figure(figsize=(8, 6))

# Plot all planets
for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]
    plt.scatter(np.log(subset["pl_bmassj"]), np.log(subset["pl_eqt"]), label=method)

# Highlight the coldest planet
plt.scatter(np.log(df2.loc[coldest_planet, "pl_bmassj"]), np.log(df2.loc[coldest_planet, "pl_eqt"]), color="red", marker="*", s=200, label=df2.loc[coldest_planet, "pl_name"])
plt.title("Planet Mass vs. Equilibrium Temperature (Log Scale)")
plt.xlabel("Log of Planet Mass")
plt.ylabel("Log of Equilibrium Temperature")
plt.legend(title="Discovery Method")
plt.show()
```
