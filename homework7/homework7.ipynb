```python
median_latitude = df["latitude"].median()
median_longitude = df["longitude"].median()
```

```python
print("Median Latitude:", median_latitude)
print("Median Longitude:", median_longitude)
```

```python
df["Latitude Distance"] = abs(df["latitude"] - median_latitude)
print(df["Latitude Distance"])
```

```python
print(df.head())
```

```python
df["Longitude Distance"] = abs(df["longitude"] - median_longitude)
print(df["Longitude Distance"])
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
closest_lat = df["Latitude Distance"].idxmin()
print(f"The state closest to the median latitude is {df.loc[closest_lat, 'state']} by {df.loc[closest_lat, 'Latitude Distance']:.2f}")
```

```python
farthest_lat = df["Latitude Distance"].idxmax()
print(f"The state farthest from the median latitude is {df.loc[farthest_lat, 'state']} by {df.loc[farthest_lat, 'Latitude Distance']:.2f}")
```

```python
closest_long = df["Longitude Distance"].idxmin()
print(f"The state closest to the median longitude is {df.loc[closest_long, 'state']} by {df.loc[closest_long, 'Longitude Distance']:.2f}")
```

```python
farthest_long = df["Longitude Distance"].idxmax()
print(f"The state farthest from the median longitude is {df.loc[farthest_long, 'state']} by {df.loc[farthest_long, 'Longitude Distance']:.2f}")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Total crashes vs speeding as top left graph
sns.scatterplot(data=df2, x="speeding", y="total", color="blue", ax=axes[0,0])
axes[0,0].set_title("Total Crashes vs Speeding")
axes[0,0].set_xlabel("Speeding (percentage)")
axes[0,0].set_ylabel("Total Crashes (per billion miles)")

# Insurance premium vs insurance losses as top right graph
sns.scatterplot(data=df2, x="ins_premium", y="ins_losses", color="green", ax=axes[0,1])
axes[0,1].set_title("Insurance Premium vs Insurance Losses")
axes[0,1].set_xlabel("Insurance Premium ($)")
axes[0,1].set_ylabel("Insurance Losses ($)")

# Histogram of total crashes as bottom left graph
sns.histplot(data=df2, x="total", color="orange", ax=axes[1,0])
axes[1,0].set_title("Distribution of Total Crashes")
axes[1,0].set_xlabel("Total Crashes (per billion miles)")
axes[1,0].set_ylabel("Number of States")

# Boxplot of total crashes by speeding level as bottom right graph
sns.boxplot(data=df2, x="abbrev", y="speeding", color="purple", ax=axes[1,1])
axes[1,1].set_title("Speeding Across States")
axes[1,1].set_xlabel("State Abbreviation")
axes[1,1].set_ylabel("Speeding (percentage)")
axes[1,1].tick_params(axis="x", rotation=90)

axes[0,0].legend(["States"])
plt.show()
```

```python
# Top left is a scatterplot that compares the percentage of drivers involved in speeding-related fatal crashes with the total number of fatal crashes. I chose these columns because I wanted to see if speeding was related to the overall number of crashes.
# Top right is a scatterplot that compares the average insurance premiums with insurance losses across states. I chose these columns because I wanted to see if states with higher insurance premiums also experience higher insurance losses.
# Bottom left is a histogram that shows the total number of fatal crashes distributed across the states. I chose these columns because I wanted to see which total crash value occurs most frequently compared between the states. 
# Bottom right is a boxplot that displays speeding-related crash percentages for each state. I chose these columns because I wanted to compare speeding-related crashes between all the states.
# I learned many things from each subplot. I was able to learn many percentages between states, as well as connections between crashes and speeding violations. This information has helped me learn a lot.
```
