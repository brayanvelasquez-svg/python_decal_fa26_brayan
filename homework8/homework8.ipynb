```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2*np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

fig, ax = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    ax[0].plot(x, make_sine_wave(x, A, 1), label='Amplitude = ' + str(A))
    ax[1].plot(x, make_sine_wave(x, 1, w), label='Frequency = ' + str(w))
    ax[2].plot(x, make_sine_wave(x, A, w), label='Amplitude = ' + str(A) + ', Frequency = ' + str(w))


ax[0].set_title('Varying Amplitude')
ax[0].set_xlabel('Input Values (x)')
ax[0].set_ylabel('Wave Amplitude (y)')
ax[0].legend()
ax[0].grid()

ax[1].set_title('Varying Frequency')
ax[1].set_xlabel('Input Values (x)')
ax[1].set_ylabel('Wave Amplitude (y)')
ax[1].legend()
ax[1].grid()

ax[2].set_title('Varying Amplitude and Frequency')
ax[2].set_xlabel('Input Values (x)')
ax[2].set_ylabel('Wave Amplitude (y)')
ax[2].legend()
ax[2].grid()

plt.show()

```
