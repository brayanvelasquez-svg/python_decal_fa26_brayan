```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
data.shape
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print("Minimum magnitude:", np.min(magnitude))
print("Maximum magnitude:", np.max(magnitude))

print("Minimum temperature:", np.min(temperature))
print("Maximum temperature:", np.max(temperature))
```

```python
print("Star with minimum magnitude:", np.argmin(magnitude))
print("Star with maximum magnitude:", np.argmax(magnitude))

print("Star with minimum temperature:", np.argmin(temperature))
print("Star with maximum temperature:", np.argmax(temperature))
```

```python
# The brightest star is 1988. The dimmest star is 264.
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, c=temperature, cmap='coolwarm', label='Stars')
plt.colorbar(label='Temperature (K)')
# Bonus colorbar is above

plt.scatter(log_temperature[np.argmin(magnitude)], log_magnitude[np.argmin(magnitude)], color='black', marker='*', s=300, label='Brightest')
plt.scatter(log_temperature[np.argmax(magnitude)], log_magnitude[np.argmax(magnitude)], color='purple', marker='*', s=300, label='Dimmest')
plt.scatter(log_temperature[np.argmax(temperature)], log_magnitude[np.argmax(temperature)], color='green', marker='*', s=300, label='Hottest')
plt.scatter(log_temperature[np.argmin(temperature)], log_magnitude[np.argmin(temperature)], color='orange', marker='*', s=300, label='Coldest')

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.title('Hertzsprung-Russel Diagram')
plt.xlabel('Temperature (log K)')
plt.ylabel('Magnitude')
plt.legend()
plt.show()
```
