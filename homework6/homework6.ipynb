```python
stars = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
def prime_stars(stars):
    rows = []
    
    for row in stars:
        for num in row:
            if num > 1:
                prime = True

                for i in range(2, num):
                    if num % i == 0:
                        prime = False

                if prime:
                    rows.append(row)
                    break
    return np.array(rows)
```

```python
print(prime_stars(stars))
```

```python
def zeros_and_ones():
    zeros = np.zeros((8, 8), dtype=int)
    
    for row in range(1, 8, 2):
        zeros[row] = np.array([1 if col % 2 == 0 else 0 for col in range(8)])
        
    return zeros
zeros_ones = zeros_and_ones()
print(zeros_ones)
```

```python
def ones_and_zeros():
    zeros = np.zeros((8, 8), dtype=int)
    
    for row in range(0, 8, 2):
        zeros[row] = np.array([1 if col % 2 == 0 else 0 for col in range(8)])
        
    return zeros
ones_zeros = ones_and_zeros()
print(ones_zeros)
```

```python
def board():
    zeros = np.zeros((8, 8), dtype=int)
    
    for row in range(8):
        if row % 2 == 0:
            zeros[row] = np.array([1 if col % 2 == 0 else 0 for col in range(8)])
        else:
            zeros[row] = np.array([0 if col % 2 == 0 else 1 for col in range(8)])
    
    return zeros

checkerboard = board()
print(checkerboard)
```

```python
def rev_board():
    zeros = np.zeros((8, 8), dtype=int)
    
    for row in range(8):
        if row % 2 == 0:
            zeros[row] = np.array([0 if col % 2 == 0 else 1 for col in range(8)])
        else:
            zeros[row] = np.array([1 if col % 2 == 0 else 0 for col in range(8)])
    
    return zeros

rev_checkerboard = rev_board()
print(rev_checkerboard)
```

```python
universe = np.array(["quasar", "black hole"])
def expansion(universe, spaces):
    result = []

    for word in universe:
        result.append((" " * spaces).join(word))
        
    return np.array(result)
universe = np.array(["quasar", "black hole"])
print(expansion(universe, 0))
print(expansion(universe, 1))
print(expansion(universe, 2))
print(expansion(universe, 3))
print(expansion(universe, 4))
print(expansion(universe, 5))
```
