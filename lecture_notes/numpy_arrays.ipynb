```python
x = np.array([1, 2, 3, 4, 5])

y1 = np.array([1, 2, 3, 4, 5])
y2 = np.array([1, 4, 9, 16, 25])
y3 = np.array([1, 8, 27, 64, 125])

plt.plot(x, y1, color="purple", label="y = x")
plt.plot(x, y2, color="blue", label="y = x squared")
plt.plot(x, y3, color="green", label="y = x cubed")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Multiple Lines")
plt.grid()
plt.legend()
plt.show
```
